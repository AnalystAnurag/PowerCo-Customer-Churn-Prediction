```python
#Let's check the consistency of features

def describe_categorical(dataset):
    cat_columns = dataset.dtypes[dataset.dtypes=='object'].index.tolist()
    if len(cat_columns)!=0:
        print('Categorical variables are', cat_columns,'\n'+'=='*40)
        for cat in cat_columns:
            describe_frame = dataset[[cat]].value_counts().reset_index(name='count')
            describe_frame['proportion'] = describe_frame['count']/len(dataset)
            print(describe_frame,'\n'+'--'*40) 
    else:
        print('There is no categorical variables in the dataset')
```

```python
# Let's check the numeric features
def describe_numeric(dataset):
    # for numeric columns whose values are discrete, display values and their frequency
    int_columns = dataset.dtypes[dataset.dtypes=='int'].index.tolist()
    # for numeric columns whose values are continuous, plot their value distribution
    float_columns = dataset.dtypes[dataset.dtypes=='float'].index.tolist()
    if len(int_columns)+len(float_columns)!=0:
        print('Numeric variables are', int_columns+float_columns,'\n'+'=='*40)
        # integer numeric feature
        if len(int_columns)!=0:    
            for cat in int_columns:
                describe_frame = dataset[[cat]].value_counts().reset_index(name='count')
                describe_frame['proportion'] = describe_frame['count']/len(dataset)
                print(describe_frame,'\n'+'--'*40)    # display value count and proportion of a categorical feature
        # 
        print(dataset[int_columns + float_columns].describe().loc[['min','max','mean','50%']].T.rename(columns={'50%':'median'}), '\n'+'--'*40)
    else:
        print('There is no numeric variables in the dataset')
```

```python
#Now, let's visualize our data
# Let's define a function to plot barchart to display features distribution with perecent labels

def plot_bar_chart_with_percent_label(df, target_variable, if_sort=False, gap_label_bar=0, figsize=(9,6)):
    # prepare data
    plot_data = client[[target_variable]].value_counts().reset_index(name='count')
    plot_data['percent'] = plot_data['count']/plot_data['count'].sum()
    if if_sort:
        x_order = plot_data.sort_values(by=['percent'], ascending=False)[target_variable]
    else:
        x_order = plot_data.sort_values(by=[target_variable], ascending=True)[target_variable]
    # plot
    fig = plt.figure(figsize=figsize)
    ax = cat.barplot(data=plot_data, x=target_variable, y='percent', order=x_order)
    # add label
    for p in ax.patches:
        x = p.get_x() + p.get_width()/2
        y = p.get_height() + gap_label_bar
        ax.annotate(text='{:.2f}%'.format(p.get_height()*100), xy=(x, y), ha='center')
    ax.margins(y=0.1)

# plot stacked bar chart with percent label by churn
def plot_stacked_bar_with_percent_label_by_churn(df, target_variable, if_sort=False, figsize=(9,6)):
    plot_data = df.groupby([target_variable,'churn']).size().reset_index(name='count')
    plot_data = plot_data.pivot(index=target_variable, columns='churn', values='count').fillna(0)
    plot_data = plot_data.div(plot_data.sum(axis=1), axis=0)
    if if_sort:
        plot_data = plot_data.sort_values(by=[0])
    ax = plot_data.plot(kind='bar', rot=False, stacked=True, figsize=figsize)
    plt.legend(['stay', 'switch'], loc='upper center', ncol=2)
    # add label
    for p in ax.patches:
        percentage = '{:.1f}%'.format(p.get_height()*100)
        if percentage!='0.0%':
            x = p.get_x() + p.get_width()/2
            y = 1 - p.get_height()/1.8
            ax.annotate(text=percentage, xy=(x, y), ha='center')
    ax.margins(y=0.1)
```

```python
# Create cv dataset
from sklearn.base import clone
# ============================================================
# 1. MODELS (NO EARLY STOPPING ANYWHERE)
# ============================================================

models = {
    'XGBoost': XGBClassifier(
        n_estimators=300,          # tuned via CV instead
        learning_rate=0.05,
        max_depth=6,
        subsample=0.8,
        colsample_bytree=0.8,
        objective='binary:logistic',
        random_state=29,
        n_jobs=-1
    ),

    'Logistic Regression': LogisticRegression(
        max_iter=1000,
        solver='liblinear',
        random_state=29
    ),

    'SVM': SVC(
        kernel='rbf',
        probability=True,
        random_state=29
    ),

    'Random Forest': RandomForestClassifier(
        n_estimators=300,
        random_state=29,
        n_jobs=-1
    )
}

# ============================================================
# 2. STORAGE
# ============================================================

pred_train_labels = {
    name: np.zeros(len(X_train_val))
    for name in models.keys()
}

cv_results = {
    name: {'precision': [], 'recall': [], 'accuracy': [], 'f1': []}
    for name in models.keys()
}

# ============================================================
# 3. CV SETUP
# ============================================================

kfold = StratifiedKFold(n_splits=5, shuffle=True, random_state=29)

print("Starting cross-validation training...")
print("=" * 80)

# ============================================================
# 4. CV LOOP
# ============================================================

for fold, (train_idx, test_idx) in enumerate(
        kfold.split(X_train_val, y_train_val), 1):

    X_train, X_test = X_train_val.iloc[train_idx], X_train_val.iloc[test_idx]
    y_train, y_test = y_train_val.iloc[train_idx], y_train_val.iloc[test_idx]

    scaler = StandardScaler()
    X_train_scaled = scaler.fit_transform(X_train)
    X_test_scaled = scaler.transform(X_test)

    print(f"\nFold {fold}")

    for model_name, base_model in models.items():

        model = clone(base_model)

        # -------- XGBOOST --------
        if model_name == 'XGBoost':
            model.fit(X_train, y_train)
            y_pred = model.predict(X_test)
            y_pred_proba = model.predict_proba(X_test)[:, 1]

        # -------- SCALED MODELS --------
        elif model_name in ['Logistic Regression', 'SVM']:
            model.fit(X_train_scaled, y_train)
            y_pred = model.predict(X_test_scaled)
            y_pred_proba = model.predict_proba(X_test_scaled)[:, 1]

        # -------- TREE MODELS --------
        else:
            model.fit(X_train, y_train)
            y_pred = model.predict(X_test)
            y_pred_proba = model.predict_proba(X_test)[:, 1]

        # Store OOF predictions
        pred_train_labels[model_name][test_idx] = y_pred_proba

        # Metrics
        precision = precision_score(y_test, y_pred)
        recall = recall_score(y_test, y_pred)
        accuracy = accuracy_score(y_test, y_pred)
        f1 = f1_score(y_test, y_pred)

        cv_results[model_name]['precision'].append(precision)
        cv_results[model_name]['recall'].append(recall)
        cv_results[model_name]['accuracy'].append(accuracy)
        cv_results[model_name]['f1'].append(f1)

        print(
            f"{model_name:20s} | "
            f"P:{precision:.3f} R:{recall:.3f} "
            f"A:{accuracy:.3f} F1:{f1:.3f}"
        )

    print("-" * 80)

print("Cross-validation completed!")
```

```python
from statsmodels.stats.contingency_tables import mcnemar

print("\n" + "="*80)
print("STATISTICAL SIGNIFICANCE TESTS (McNemar) — XGBoost as Reference")
print("="*80)

# ----------------------------------------------------
# Use OUT-OF-FOLD labels (CRITICAL)
# ----------------------------------------------------
y_true = y_oof # must match pred_train_labels length

# Convert predictions to hard labels if needed
all_predictions = {}
for model_name in models.keys():
    preds = pred_train_labels[model_name]

    # If probabilities accidentally stored
    if preds.ndim > 1:
        preds = (preds[:, 1] >= 0.5).astype(int)

    all_predictions[model_name] = preds

xgb_predictions = all_predictions['XGBoost']

# ----------------------------------------------------
# McNemar Tests: XGBoost vs Others
# ----------------------------------------------------
for model_name in ['Random Forest', 'Logistic Regression', 'SVM']:

    other_predictions = all_predictions[model_name]
    table = np.zeros((2, 2))

    for i in range(len(y_true)):
        if xgb_predictions[i] == y_true[i] and other_predictions[i] == y_true[i]:
            table[0, 0] += 1
        elif xgb_predictions[i] != y_true[i] and other_predictions[i] == y_true[i]:
            table[0, 1] += 1
        elif xgb_predictions[i] == y_true[i] and other_predictions[i] != y_true[i]:
            table[1, 0] += 1
        else:
            table[1, 1] += 1

    result = mcnemar(table, exact=False, correction=True)

    print(f"\nMcNemar's test: XGBoost vs {model_name}")
    print("Contingency Table:")
    print(table)
    print(f"Statistic: {result.statistic:.4f}, p-value: {result.pvalue:.6f}")

    if result.pvalue < 0.05:
        print("✓ Statistically SIGNIFICANT difference (p < 0.05)")
        if table[1, 0] > table[0, 1]:
            print("  ➤ XGBoost performs SIGNIFICANTLY BETTER")
        else:
            print("  ➤ XGBoost performs SIGNIFICANTLY WORSE")
    else:
        print("✗ No statistically significant difference")
```

```python
from statsmodels.stats.contingency_tables import mcnemar

# --- Hard labels using tuned threshold ---
xgb_pred_opt = (xgb_proba >= best_threshold).astype(int)

# Fix: Access rf_proba as a 1D array since pred_train_labels['Random Forest'] is 1D
rf_proba = pred_train_labels['Random Forest']  # Remove the [:, 1] indexing
rf_pred = (rf_proba >= 0.5).astype(int)  # RF usually OK at 0.5

# --- Build McNemar contingency table ---
table = np.zeros((2, 2))

for i in range(len(y_oof)):
    xgb_correct = (xgb_pred_opt[i] == y_oof[i])
    rf_correct = (rf_pred[i] == y_oof[i])

    if xgb_correct and rf_correct:
        table[0, 0] += 1
    elif not xgb_correct and rf_correct:
        table[0, 1] += 1
    elif xgb_correct and not rf_correct:
        table[1, 0] += 1
    else:
        table[1, 1] += 1

# --- McNemar test ---
result = mcnemar(table, exact=False)

print("\n" + "="*80)
print("MCNEMAR TEST (XGBoost vs Random Forest — Tuned Threshold)")
print("="*80)
print("Contingency Table:")
print(table)
print(f"Statistic : {result.statistic:.4f}")
print(f"p-value   : {result.pvalue:.6f}")

if result.pvalue < 0.05:
    print("✓ Statistically significant difference")
    if table[1, 0] > table[0, 1]:
        print("➤ XGBoost performs SIGNIFICANTLY BETTER")
    else:
        print("➤ Random Forest performs SIGNIFICANTLY BETTER")
else:
    print("✗ No statistically significant difference")
```

### 1. Model Performance & Selection

* Multiple machine learning models were evaluated using stratified cross-validation, including Logistic Regression, SVM, Random Forest, and XGBoost.

* While accuracy remained consistently high across models (~90%), this metric alone was misleading due to class imbalance in churn outcomes.

* XGBoost and Random Forest emerged as the top-performing models in terms of precision–recall trade-off and F1-score, outperforming linear and margin-based classifiers.

* A McNemar statistical significance test confirmed that performance differences were not due to random variation, strengthening the validity of model comparison.

### 2. Class Imbalance & Metric Interpretation

* Extremely low recall values across several models indicate a highly imbalanced dataset, where churn events are rare.

* Precision showed high variance across folds, especially for Random Forest, highlighting sensitivity to data splits.

* Threshold tuning significantly altered model behavior, reinforcing that default probability cut-offs are inappropriate for churn problems.

### 3. Statistical Validation (McNemar Test)

* McNemar’s test demonstrated statistically significant differences between competing models.

* Under-tuned decision thresholds, Random Forest outperformed XGBoost, despite XGBoost showing competitive average cross-validation metrics.

* This result highlights that mean CV performance alone is insufficient without statistical testing.

### 4. Feature Importance & Explainability (SHAP Analysis)

* SHAP analysis provided both global and local interpretability, identifying the most influential predictors driving churn.

* A small subset of variables consistently contributed the majority of predictive power, indicating feature concentration rather than diffuse influence.

* The directionality of SHAP values revealed non-linear and interaction effects, which are not captured by linear models.

* The top 10 features driving churn are :

   *gross margin on power subscription(margin_gross_pow_ele)

   *subscribed power(pow_max)

   *code of the electricity campaign the customer first subscribed to(origin_up)

   *total net margin(net_margin)

   *electricity consumption of the past 12 months(cons_12m)

   *electricity consumption of the last month(cons_last_month)

   *diff_dec_jan_price_off_peak_var

   *forecasted bill of meter rental for the next 2 months(forecast_meter_rent_12m)

   *antiquity of the client (in number of years)(num_years_antig)

   *forecasted electricity consumption for next 12 months(forecast_cons_12m)
